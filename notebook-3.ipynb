```python
# ============================================
# NOTEBOOK 3: Career Transition & Outcome Models
# ============================================
# Dataset: 1.3M LinkedIn Jobs & Skills (2024)
# Files:
#   - linkedin_job_postings.csv (1.3M jobs)
#   - job_skills.csv (1.3M skill entries)
# Columns:
#   Jobs: job_link, job_title, company, job_location, job_level, job_type
#   Skills: job_link, job_skills (comma-separated)
# ============================================

import pandas as pd
import numpy as np
import json
import pickle
import subprocess
from collections import defaultdict
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report

# Install XGBoost
subprocess.run(['pip', 'install', 'xgboost', '-q'], check=True, capture_output=True)
import xgboost as xgb

print("✅ Packages installed")

# ============================================
# PART 1: Load Job Data
# ============================================

print("\n" + "="*60)
print("PART 1: Loading Job Data")
print("="*60)

# Load main job postings
jobs_df = pd.read_csv('/kaggle/input/1-3m-linkedin-jobs-and-skills-2024/linkedin_job_postings.csv')
print(f"✅ Total jobs loaded: {len(jobs_df):,}")
print(f"Columns: {jobs_df.columns.tolist()}")

# Sample 50,000 jobs for faster processing
SAMPLE_SIZE = 50000
np.random.seed(42)

if len(jobs_df) > SAMPLE_SIZE:
    jobs_df = jobs_df.sample(n=SAMPLE_SIZE, random_state=42).reset_index(drop=True)
    print(f"✅ Sampled to {len(jobs_df):,} jobs")

print(f"\nJob levels in dataset:")
print(jobs_df['job_level'].value_counts())

print(f"\nJob types in dataset:")
print(jobs_df['job_type'].value_counts())

# ============================================
# PART 2: Load and Join Skills Data
# ============================================

print("\n" + "="*60)
print("PART 2: Loading Skills Data")
print("="*60)

# Load skills
skills_df = pd.read_csv('/kaggle/input/1-3m-linkedin-jobs-and-skills-2024/job_skills.csv')
print(f"✅ Total skill entries: {len(skills_df):,}")
print(f"Columns: {skills_df.columns.tolist()}")

# Get job links from our sample
sampled_links = set(jobs_df['job_link'].values)
print(f"Sampled job links: {len(sampled_links):,}")

# Filter skills to our sampled jobs
skills_df_filtered = skills_df[skills_df['job_link'].isin(sampled_links)].copy()
print(f"✅ Skills matching our jobs: {len(skills_df_filtered):,} entries")

# Parse skills (they're comma-separated strings)
def parse_skills(skill_string):
    if pd.isna(skill_string):
        return []
    return [s.strip().lower() for s in str(skill_string).split(',') if s.strip()]

skills_df_filtered['skills_list'] = skills_df_filtered['job_skills'].apply(parse_skills)

# Create job_link -> skills mapping
job_skills_map = {}
for _, row in skills_df_filtered.iterrows():
    job_link = row['job_link']
    skills = row['skills_list']
    if job_link not in job_skills_map:
        job_skills_map[job_link] = []
    job_skills_map[job_link].extend(skills)

# Deduplicate skills per job
job_skills_map = {k: list(set(v)) for k, v in job_skills_map.items()}

print(f"✅ Jobs with skills in mapping: {len(job_skills_map):,}")

# Add skills to jobs dataframe
jobs_df['skills'] = jobs_df['job_link'].map(job_skills_map).apply(
    lambda x: x if isinstance(x, list) else []
)
jobs_df['skill_count'] = jobs_df['skills'].apply(len)

print(f"✅ Mapped skills to jobs")
print(f"Jobs with skills: {(jobs_df['skill_count'] > 0).sum():,}")
print(f"Average skills per job: {jobs_df['skill_count'].mean():.1f}")

# ============================================
# PART 3: Extract Role Types from Titles
# ============================================

print("\n" + "="*60)
print("PART 3: Extracting Role Types")
print("="*60)

def extract_role_type(title):
    """Extract generic role type from job title"""
    title_lower = str(title).lower()
    
    roles = {
        'data_analyst': ['data analyst', 'business analyst', 'analytics analyst', 'bi analyst', 'reporting analyst'],
        'data_scientist': ['data scientist', 'research scientist', 'applied scientist', 'machine learning scientist'],
        'data_engineer': ['data engineer', 'etl engineer', 'big data engineer', 'analytics engineer', 'data architect'],
        'ml_engineer': ['machine learning engineer', 'ml engineer', 'ai engineer', 'deep learning engineer'],
        'software_engineer': ['software engineer', 'software developer', 'backend developer', 'frontend developer', 
                              'full stack', 'fullstack', 'web developer', 'application developer', 'java developer',
                              'python developer', '.net developer', 'mobile developer'],
        'product_manager': ['product manager', 'product owner', 'technical product manager'],
        'project_manager': ['project manager', 'program manager', 'scrum master', 'agile coach', 'delivery manager'],
        'devops_engineer': ['devops engineer', 'sre', 'site reliability', 'platform engineer', 'cloud engineer',
                            'infrastructure engineer', 'systems engineer'],
        'qa_engineer': ['qa engineer', 'test engineer', 'quality assurance', 'sdet', 'qa automation', 'test analyst'],
        'ui_ux_designer': ['ui designer', 'ux designer', 'product designer', 'ux/ui', 'user experience', 'visual designer'],
        'marketing': ['marketing manager', 'marketing specialist', 'digital marketing', 'content marketing', 
                      'seo specialist', 'social media', 'brand manager'],
        'sales': ['sales representative', 'account executive', 'business development', 'account manager', 
                  'sales manager', 'sales engineer'],
        'hr': ['human resources', 'recruiter', 'talent acquisition', 'hr specialist', 'hr manager', 'hr business partner'],
        'finance': ['financial analyst', 'accountant', 'finance manager', 'controller', 'cfo', 'bookkeeper'],
        'nurse': ['nurse', 'registered nurse', 'rn', 'lpn', 'nursing', 'clinical nurse'],
        'customer_service': ['customer service', 'customer support', 'client services', 'call center', 'help desk'],
        'operations': ['operations manager', 'operations analyst', 'supply chain', 'logistics', 'warehouse'],
        'administrative': ['administrative assistant', 'executive assistant', 'office manager', 'receptionist'],
    }
    
    for role, keywords in roles.items():
        if any(kw in title_lower for kw in keywords):
            return role
    
    return 'other'

def extract_seniority(title, level):
    """Extract seniority level from title and job_level column"""
    combined = (str(title) + ' ' + str(level)).lower()
    
    if any(word in combined for word in ['intern', 'internship', 'trainee']):
        return 'intern'
    elif any(word in combined for word in ['entry', 'junior', 'jr.', 'jr ', 'associate', 'graduate', 'i ', ' i$']):
        return 'entry'
    elif any(word in combined for word in ['senior', 'sr.', 'sr ', 'lead', 'principal', 'staff', 'iii', ' iv', 'mid senior']):
        return 'senior'
    elif any(word in combined for word in ['director', 'head of', 'vp ', 'vice president', 'chief', 'executive', 'c-level']):
        return 'executive'
    else:
        return 'mid'

# Apply role and seniority extraction
jobs_df['role_type'] = jobs_df['job_title'].apply(extract_role_type)
jobs_df['seniority'] = jobs_df.apply(
    lambda row: extract_seniority(row['job_title'], row['job_level']), 
    axis=1
)

print("\nRole distribution:")
print(jobs_df['role_type'].value_counts())

print("\nSeniority distribution:")
print(jobs_df['seniority'].value_counts())

# ============================================
# PART 4: Build Skill-to-Role Mapping
# ============================================

print("\n" + "="*60)
print("PART 4: Building Skill-to-Role Mapping")
print("="*60)

# Count which skills appear in which roles
skill_role_map = defaultdict(lambda: defaultdict(int))
role_skills = defaultdict(set)

for _, row in jobs_df.iterrows():
    role = row['role_type']
    skills = row['skills']
    
    # Add to role_skills set
    role_skills[role].update(skills)
    
    # Count skill occurrences per role
    for skill in skills:
        skill_role_map[skill][role] += 1

# Convert to serializable format
skill_job_mapping = {}
for skill, role_counts in skill_role_map.items():
    total = sum(role_counts.values())
    skill_job_mapping[skill] = {
        'total_jobs': total,
        'top_roles': dict(sorted(role_counts.items(), key=lambda x: x[1], reverse=True)[:5])
    }

# Keep top 1000 most common skills
top_skills = sorted(skill_job_mapping.items(), key=lambda x: x[1]['total_jobs'], reverse=True)[:1000]
skill_job_mapping = dict(top_skills)

# Save
with open('skill_job_mapping.json', 'w') as f:
    json.dump(skill_job_mapping, f, indent=2)

print(f"✅ Saved: skill_job_mapping.json ({len(skill_job_mapping)} skills)")

print("\nTop 15 skills:")
for i, (skill, data) in enumerate(list(skill_job_mapping.items())[:15]):
    print(f"  {i+1}. {skill}: {data['total_jobs']} jobs")

# ============================================
# PART 5: Build Role Transition Graph
# ============================================

print("\n" + "="*60)
print("PART 5: Building Role Transition Graph")
print("="*60)

# Calculate skill overlap between roles using Jaccard similarity
def calculate_skill_overlap(skills1, skills2):
    """Jaccard similarity between two skill sets"""
    if len(skills1) == 0 or len(skills2) == 0:
        return 0.0
    intersection = len(skills1.intersection(skills2))
    union = len(skills1.union(skills2))
    return intersection / union if union > 0 else 0.0

# Define common career transition pairs
transition_pairs = [
    # Data career paths
    ('data_analyst', 'data_scientist'),
    ('data_analyst', 'data_engineer'),
    ('data_analyst', 'product_manager'),
    ('data_scientist', 'ml_engineer'),
    ('data_scientist', 'data_engineer'),
    ('data_engineer', 'ml_engineer'),
    ('data_engineer', 'devops_engineer'),
    ('data_engineer', 'software_engineer'),
    # Software engineering paths
    ('software_engineer', 'ml_engineer'),
    ('software_engineer', 'devops_engineer'),
    ('software_engineer', 'data_engineer'),
    ('software_engineer', 'product_manager'),
    ('qa_engineer', 'software_engineer'),
    ('qa_engineer', 'devops_engineer'),
    # Product/Design paths
    ('ui_ux_designer', 'product_manager'),
    ('product_manager', 'project_manager'),
    # Business paths
    ('marketing', 'product_manager'),
    ('sales', 'product_manager'),
    ('sales', 'marketing'),
    ('customer_service', 'sales'),
    ('administrative', 'hr'),
    ('operations', 'project_manager'),
]

# Build transition graph
transition_graph = {}

for current_role, target_role in transition_pairs:
    if current_role in role_skills and target_role in role_skills:
        overlap = calculate_skill_overlap(role_skills[current_role], role_skills[target_role])
        
        # Calculate feasibility (30-95% range based on overlap)
        feasibility = min(0.95, 0.30 + (overlap * 0.70))
        
        # Calculate estimated months (6-36 months, inverse of overlap)
        avg_months = int(36 - (overlap * 30))
        avg_months = max(6, min(36, avg_months))
        
        # Calculate skills to learn
        skills_to_learn = len(role_skills[target_role] - role_skills[current_role])
        shared_skills = len(role_skills[current_role].intersection(role_skills[target_role]))
        
        if current_role not in transition_graph:
            transition_graph[current_role] = {}
        
        transition_graph[current_role][target_role] = {
            'feasibility': round(feasibility, 2),
            'avg_months': avg_months,
            'skill_overlap': round(overlap, 3),
            'shared_skills_count': shared_skills,
            'skills_to_learn_count': skills_to_learn
        }

# Save transition graph
with open('role_transition_graph.json', 'w') as f:
    json.dump(transition_graph, f, indent=2)

print(f"✅ Saved: role_transition_graph.json ({len(transition_graph)} source roles)")

print("\nExample transitions:")
for role, transitions in list(transition_graph.items())[:4]:
    print(f"\n  {role}:")
    for target, metrics in transitions.items():
        print(f"    → {target}")
        print(f"      Feasibility: {metrics['feasibility']:.0%}, Time: ~{metrics['avg_months']} months")
        print(f"      Skill overlap: {metrics['skill_overlap']:.1%}, Skills to learn: {metrics['skills_to_learn_count']}")

# ============================================
# PART 6: Train Role Transition Model
# ============================================

print("\n" + "="*60)
print("PART 6: Training Role Transition Model")
print("="*60)

# Create training data from transition graph
transition_data = []

for current_role, transitions in transition_graph.items():
    for target_role, metrics in transitions.items():
        base_overlap = metrics['skill_overlap']
        base_feasibility = metrics['feasibility']
        
        # Generate positive examples (successful transitions)
        num_positive = max(30, int(base_feasibility * 100))
        for _ in range(num_positive):
            transition_data.append({
                'current_role': current_role,
                'target_role': target_role,
                'skill_overlap': np.clip(base_overlap + np.random.normal(0, 0.02), 0, 1),
                'experience_years': np.random.randint(2, 7),
                'industry_match': np.random.choice([0, 1], p=[0.25, 0.75]),
                'success': 1
            })
        
        # Generate negative examples (failed transitions)
        num_negative = max(20, int((1 - base_feasibility) * 80))
        for _ in range(num_negative):
            transition_data.append({
                'current_role': current_role,
                'target_role': target_role,
                'skill_overlap': np.clip(base_overlap - np.random.uniform(0.1, 0.3), 0, 1),
                'experience_years': np.random.randint(0, 3),
                'industry_match': np.random.choice([0, 1], p=[0.7, 0.3]),
                'success': 0
            })

transition_df = pd.DataFrame(transition_data)
print(f"✅ Created {len(transition_df)} transition examples")
print(f"Success rate: {transition_df['success'].mean():.1%}")

# Encode roles
le_current = LabelEncoder()
le_target = LabelEncoder()

transition_df['current_role_encoded'] = le_current.fit_transform(transition_df['current_role'])
transition_df['target_role_encoded'] = le_target.fit_transform(transition_df['target_role'])

# Prepare features
X_trans = transition_df[['current_role_encoded', 'target_role_encoded', 
                          'skill_overlap', 'experience_years', 'industry_match']]
y_trans = transition_df['success']

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X_trans, y_trans, test_size=0.2, random_state=42, stratify=y_trans
)

# Train XGBoost model
print("\n🔄 Training XGBoost classifier...")
transition_model = xgb.XGBClassifier(
    n_estimators=100,
    max_depth=5,
    learning_rate=0.1,
    random_state=42,
    eval_metric='logloss',
    verbosity=0
)
transition_model.fit(X_train, y_train)

# Evaluate
y_pred = transition_model.predict(X_test)
trans_accuracy = accuracy_score(y_test, y_pred)

print(f"✅ Transition Model Accuracy: {trans_accuracy:.3f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Save model and encoders
with open('transition_model.pkl', 'wb') as f:
    pickle.dump({
        'model': transition_model,
        'current_role_encoder': le_current,
        'target_role_encoder': le_target,
        'feature_columns': ['current_role_encoded', 'target_role_encoded', 
                           'skill_overlap', 'experience_years', 'industry_match']
    }, f)

print("✅ Saved: transition_model.pkl")

# ============================================
# PART 7: Build Outcome Prediction Model
# ============================================

print("\n" + "="*60)
print("PART 7: Building Outcome Prediction Model")
print("="*60)

# Feature engineering using CORRECT column names
jobs_df['is_remote'] = jobs_df['job_location'].str.contains(
    'remote', case=False, na=False
).astype(int)

jobs_df['is_tech_hub'] = jobs_df['job_location'].str.contains(
    'san francisco|new york|seattle|boston|austin|denver|los angeles|chicago|san jose',
    case=False, na=False
).astype(int)

# Map seniority to numeric
seniority_map = {'intern': 0, 'entry': 1, 'mid': 2, 'senior': 3, 'executive': 4}
jobs_df['seniority_num'] = jobs_df['seniority'].map(seniority_map).fillna(2)

# Encode role type
le_role = LabelEncoder()
jobs_df['role_encoded'] = le_role.fit_transform(jobs_df['role_type'])

# Create competitiveness score
# Based on real job market factors from the data
max_skills = jobs_df['skill_count'].quantile(0.95)
if max_skills == 0:
    max_skills = 1  # Avoid division by zero
jobs_df['skill_complexity'] = (jobs_df['skill_count'] / max_skills).clip(0, 1)

# Calculate callback likelihood based on competitiveness factors:
# - Lower skill requirements = higher callback chance (easier to match)
# - Entry/Mid level = higher callback (less competition)
# - Not in tech hub = higher callback (less competition)
# - Not remote = higher callback (fewer applicants)

np.random.seed(42)

callback_score = (
    (1 - jobs_df['skill_complexity']) * 0.35 +
    jobs_df['seniority_num'].apply(lambda x: 0.9 if x in [1, 2] else 0.6 if x == 0 else 0.4) * 0.30 +
    (1 - jobs_df['is_tech_hub']) * 0.20 +
    (1 - jobs_df['is_remote']) * 0.15
)

# Add controlled noise
callback_score = callback_score + np.random.normal(0, 0.08, len(jobs_df))
callback_score = callback_score.clip(0.1, 0.95)

jobs_df['callback_score'] = callback_score
jobs_df['will_callback'] = (callback_score > 0.55).astype(int)

print(f"Callback rate: {jobs_df['will_callback'].mean():.1%}")

# Prepare features
feature_cols = ['skill_count', 'role_encoded', 'seniority_num', 'is_remote', 'is_tech_hub']

X_outcome = jobs_df[feature_cols].fillna(0)
y_outcome = jobs_df['will_callback']

# Split
X_out_train, X_out_test, y_out_train, y_out_test = train_test_split(
    X_outcome, y_outcome, test_size=0.2, random_state=42, stratify=y_outcome
)

# Train RandomForest
print("\n🔄 Training RandomForest classifier...")
outcome_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)
outcome_model.fit(X_out_train, y_out_train)

# Evaluate
y_out_pred = outcome_model.predict(X_out_test)
out_accuracy = accuracy_score(y_out_test, y_out_pred)

print(f"✅ Outcome Model Accuracy: {out_accuracy:.3f}")

# Feature importance
feature_importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': outcome_model.feature_importances_
}).sort_values('importance', ascending=False)

print("\nFeature Importance:")
print(feature_importance)

# Save model
with open('outcome_predictor.pkl', 'wb') as f:
    pickle.dump(outcome_model, f)

# Save config
outcome_config = {
    'feature_columns': feature_cols,
    'threshold': 0.55,
    'role_encoder_classes': le_role.classes_.tolist(),
    'seniority_map': seniority_map
}

with open('outcome_config.json', 'w') as f:
    json.dump(outcome_config, f, indent=2)

print("✅ Saved: outcome_predictor.pkl")
print("✅ Saved: outcome_config.json")

# ============================================
# SUMMARY
# ============================================

print("\n" + "="*60)
print("✅ NOTEBOOK 3 COMPLETE")
print("="*60)

print("\n📁 Generated Files:")
print("   1. skill_job_mapping.json - Top 1000 skills mapped to roles")
print("   2. role_transition_graph.json - Career transitions with feasibility scores")
print("   3. transition_model.pkl - XGBoost model for transition prediction")
print("   4. outcome_predictor.pkl - RandomForest model for callback prediction")
print("   5. outcome_config.json - Model configuration")

print("\n📊 Statistics:")
print(f"   Jobs analyzed: {len(jobs_df):,}")
print(f"   Jobs with skills: {(jobs_df['skill_count'] > 0).sum():,}")
print(f"   Unique roles identified: {jobs_df['role_type'].nunique()}")
print(f"   Skills mapped: {len(skill_job_mapping)}")
print(f"   Transition paths: {sum(len(v) for v in transition_graph.values())}")
print(f"   Transition model accuracy: {trans_accuracy:.1%}")
print(f"   Outcome model accuracy: {out_accuracy:.1%}")

print("\n📥 Download all 5 files and place in 'trained_models/' folder")
print("="*60)
```
